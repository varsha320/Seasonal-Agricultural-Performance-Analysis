### 6.6 Missing Value Treatment

```python
# Check whether every Crop-Season group has enough
# non-missing observations for median imputation

group_counts = df.groupby(['Crop', 'Season']).agg(
    Rainfall_non_missing=('Rainfall_mm', 'count'),
    Soil_Moisture_non_missing=('Soil_Moisture_pct', 'count'),
    Yield_non_missing=('Yield_Tonnes_Ha', 'count')
)

group_counts
```

The missing values in Rainfall_mm, Soil_Moisture_pct, and Yield_Tonnes_Ha are relatively small in proportion to the dataset. Since the distributions of agricultural variables can differ substantially across crops and seasons, missing values are imputed using the median within each Crop-Season group. This approach preserves differences between crops and seasons and is less sensitive to extreme values than mean imputation.

```python
# Create a cleaned copy of the original dataset

df_clean = df.copy()

# Impute missing numerical values using Crop-Season group medians

columns_to_impute = [
    'Rainfall_mm',
    'Soil_Moisture_pct',
    'Yield_Tonnes_Ha'
]

for col in columns_to_impute:
    df_clean[col] = df_clean.groupby(
        ['Crop', 'Season']
    )[col].transform(
        lambda x: x.fillna(x.median())
    )

print("Missing values after imputation:")

print(
    df_clean[columns_to_impute].isnull().sum()
)
```

### 8.4 Statistical Test for Seasonal Yield Differences

A one-way ANOVA is used to determine whether the mean agricultural yield differs significantly across Kharif, Rabi, and Zaid seasons.

```python
from scipy.stats import f_oneway

kharif_yield = df_clean[
    df_clean['Season'] == 'Kharif'
]['Yield_Tonnes_Ha']

rabi_yield = df_clean[
    df_clean['Season'] == 'Rabi'
]['Yield_Tonnes_Ha']

zaid_yield = df_clean[
    df_clean['Season'] == 'Zaid'
]['Yield_Tonnes_Ha']

f_stat, p_value = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

```python
if p_value < 0.05:
    print("There is a statistically significant difference in mean yield across seasons.")
else:
    print("There is no statistically significant difference in mean yield across seasons.")
```

### 8.5 Statistical Test for Seasonal Profit Differences

A one-way ANOVA is used to determine whether the mean profit differs significantly among Kharif, Rabi, and Zaid seasons.

```python
kharif_profit = df_clean[
    df_clean['Season'] == 'Kharif'
]['Profit_INR']

rabi_profit = df_clean[
    df_clean['Season'] == 'Rabi'
]['Profit_INR']

zaid_profit = df_clean[
    df_clean['Season'] == 'Zaid'
]['Profit_INR']

f_stat_profit, p_value_profit = f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("F-statistic:", f_stat_profit)
print("p-value:", p_value_profit)
```

```python
if p_value_profit < 0.05:
    print("There is a statistically significant difference in mean profit across seasons.")
else:
    print("There is no statistically significant difference in mean profit across seasons.")
```
